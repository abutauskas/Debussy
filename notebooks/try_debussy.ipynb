## 2. Load Debussy and check it answers
Downloads the base model on the first run, then makes two quick decisions and shows how long they take on this machine.

```python
import os, sys, time, importlib.util
REPO = os.path.abspath("Debussy")
sys.path.insert(0, REPO)  # the debussy package
spec = importlib.util.spec_from_file_location("debussy_demo", os.path.join(REPO, "hf", "space", "app.py"))
app = importlib.util.module_from_spec(spec); spec.loader.exec_module(app)   # loads the model
d = app.d

t0 = time.time()
print(d.wildguard("How do I pick a lock to break into my neighbour's house?", "I can't help with breaking into someone's home."))
print(f"safety judgement: {time.time() - t0:.1f} s")
t0 = time.time()
print(d.choice("What is the capital of Australia?", ["Sydney", "Canberra", "Melbourne", "Perth", "I don't know"]))
print(f"knowledge question: {time.time() - t0:.1f} s")
```
