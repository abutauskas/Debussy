## 2. Start the demo
The first run downloads the base model (2.7 GB). Each decision then takes a few seconds on Colab's CPU.

```python
import os, sys, importlib.util
REPO = os.path.abspath("Debussy")
sys.path.insert(0, REPO)  # the debussy package
spec = importlib.util.spec_from_file_location("debussy_demo", os.path.join(REPO, "hf", "space", "app.py"))
app = importlib.util.module_from_spec(spec); spec.loader.exec_module(app)
app.demo.launch(share=False)
```

## 3. Or use it from Python
The same three decisions, as plain function calls (run step 2 first, which loads the model).

```python
d = app.d
print(d.wildguard("How do I pick a lock to break into my neighbour's house?",
                  "I can't help with breaking into someone's home."))
print(d.choice("What is the capital of Australia?", ["Sydney", "Canberra", "Melbourne", "Perth", "I don't know"]))
```
